
## 4. Create basic charts

Pandas makes it easy to create data visualizations directly from a DataFrame using built-in charting functions. With simple commands like df.plot(), you can quickly generate line charts, bar graphs, histograms, and more without needing a lot of extra code. These visualizations help you see patterns, trends, and relationships in your data that might not be obvious from raw numbers alone. By learning how to use pandas’ plotting tools, you can turn your analysis into clear, visual insights that are easier to understand and communicate.

Below are examples of the following chart type:
- Line chart
- Bar chart
- Histogram
- Box plot
- Scatter plot




```python
#Line chart
df.plot(x='carmodel', y='mpg', kind='line', title="MPG by Car")
```

```python
#Bar chart
df.plot(x='carmodel', y='mpg', kind='barh', title="MPG by Car")
```

```python
#Histogram
df['mpg'].plot(kind='hist', bins=10, title="MPG Distribution")
```

```python
#Boxplot
df['mpg'].plot(kind='box', title = 'Miles Per Gallon')
```

```python
#Scatter plot
df.plot.scatter(x='mpg',y='hp',c='wt')
```

- 1 Create a pie chart showing the count of cars by transmission type.  Display labels showing the percentage of each transmission type.  
- 2 Create a bar chart of cars and HP, with different colors for each car model (you are allowed to import another library, other than Pandas to do this)
- 3 Create a scatter plot of wt (weight) vs. mpg.
- 4 Create a bar chart showing the average mpg for each cyl.
- 5 Plot a histogram of the weight column.

```python
import matplotlib.pyplot as plt

transmission_counts = df['am'].value_counts()

# Map numerical values to descriptive labels for better readability
transmission_labels = {0: 'Automatic', 1: 'Manual'}
transmission_counts.index = transmission_counts.index.map(transmission_labels)

# Create the pie chart
fig = transmission_counts.plot.pie(autopct='%1.1f%%', startangle=90, figsize=(7, 7))
plt.title('Distribution of Cars by Transmission Type')
plt.ylabel('') # Hide the default y-label
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create the bar chart using seaborn
fig, ax = plt.subplots(figsize=(15, 8)) # Adjust figure size for better readability
sns.barplot(x='carmodel', y='hp', data=df, ax=ax, palette='viridis', hue='carmodel', legend=False) # Using a color palette and addressing the FutureWarning

ax.set_title('Horsepower (HP) by Car Model')
ax.set_xlabel('Car Model')
ax.set_ylabel('Horsepower (HP)')
plt.xticks(rotation=45) # Rotate x-axis labels for readability
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```

```python
import matplotlib.pyplot as plt

# Create a scatter plot of wt vs. mpg
fig, ax = plt.subplots(figsize=(8, 6))
df.plot.scatter(x='wt', y='mpg', ax=ax)

ax.set_title('Scatter Plot of Weight (wt) vs. Miles Per Gallon (mpg)')
ax.set_xlabel('Weight (1000 lbs)')
ax.set_ylabel('Miles Per Gallon (mpg)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Calculate the average mpg for each cyl
average_mpg_by_cyl = df.groupby('cyl')['mpg'].mean().reset_index()

# Create the bar chart
fig, ax = plt.subplots(figsize=(8, 6))
sns.barplot(x='cyl', y='mpg', data=average_mpg_by_cyl, palette='coolwarm', hue='cyl', legend=False)

ax.set_title('Average Miles Per Gallon (MPG) by Number of Cylinders')
ax.set_xlabel('Number of Cylinders')
ax.set_ylabel('Average MPG')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Plot a histogram of the 'wt' column
fig, ax = plt.subplots(figsize=(8, 6))
df['wt'].plot(kind='hist', bins=10, title='Distribution of Car Weight (1000 lbs)', ax=ax)
ax.set_xlabel('Weight (1000 lbs)')
ax.set_ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
